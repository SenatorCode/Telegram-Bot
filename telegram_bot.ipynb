```python
api_id = "26413902"
api_hash = "ce5713e8d8a6bffa69d695ed7d9fa55c"
api_key = "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"

bot_token_key = "7772319338:AAEACAfpx3EccKXi56vUdMWRTBJoC2PZA74"



import nest_asyncio
nest_asyncio.apply()
import logging
import asyncio
from telethon import TelegramClient, events
from openai import OpenAI
from telethon.sessions import MemorySession

logging.basicConfig(level=logging.INFO)

user = TelegramClient(MemorySession(),api_id,api_hash)

ai_client = OpenAI(
    base_url="http://192.168.32.1:1234/v1",  # Use the IP shown in LM Studio
    api_key="lm-studio"  # dummy key, still required
)

async def main():
    await user.start(bot_token=bot_token_key)

    
    @user.on(events.NewMessage(pattern="/start"))
    async def start_handler(event):
        await event.respond("Hello i am KMF Genie Bot, how can i help you today.")
        logging.info(f"start command received from {event.sender_id}")
        
    @user.on(events.NewMessage(pattern="/help"))                                                                                        
    async def helphandler(event):
        help_text = (
            "Here are the commands you can use:\n"
            "/start : To start the bot\n"
            "/help : To get help"
            )
        await event.respond(help_text)
        logging.info(f"Help command received from {event.sender_id}")

        

@user.on(events.NewMessage)
async def gen_responder(event):
    message = event.text.strip()
    if message.startswith("/"):
        return  

    try:
        response = ai_client.chat.completions.create(
            model="tinyllama-1.1b-chat-v1.0",
            messages=[{"role": "user", "content": message}],
            max_tokens=150
        )

        # handle both dict and object cases
        msg = response.choices[0].message
        response_text = msg["content"] if isinstance(msg, dict) else msg.content

        if response_text:
            await event.respond(response_text, reply_to=event.id)
            logging.info(f"AI replied: {response_text}")
    except Exception as e:
        logging.error(f"Error generating AI response: {e}")
        await event.respond("⚠️ Sorry, something went wrong with AI response.")

    print("🤖 KMF Genie Bot is running...")
if __name__ == "__main__":
    asyncio.run(main())
```
